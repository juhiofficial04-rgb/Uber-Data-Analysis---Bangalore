4️⃣ Exploratory Data Analysis (EDA) -

```python
uber_raw_data["Date"] = pd.to_datetime(uber_raw_data["Date"],format="%Y-%m-%d")
uber_raw_data["Time"] = pd.to_datetime(uber_raw_data["Time"], format="%H:%M:%S")
```

```python
uber_raw_data["WeekDay"] = uber_raw_data["Date"].dt.weekday
uber_raw_data["Day"] = uber_raw_data["Date"].dt.date
uber_raw_data["Hour"] = uber_raw_data["Time"].dt.hour
```

```python
# Ride Statistics
Total_Successful_Rides = uber_raw_data["BookingStatus"].value_counts()
Total_Unique_Customers = uber_raw_data["CustomerID"].value_counts()
Total_Pickup_Locations = uber_raw_data["PickupLocation"].value_counts()
Total_Vehicle_Types = uber_raw_data["VehicleType"].value_counts()
```

- Top Pickup Locations vs Ride Demand

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter

pareto_data = uber_raw_data.groupby("PickupLocation")["BookingID"].count().reset_index(name="BookingCount")

pareto_data = pareto_data.sort_values(by="BookingCount", ascending=False).reset_index(drop=True)

pareto_data["CumulativePercentage"] = (pareto_data["BookingCount"].cumsum() / pareto_data["BookingCount"].sum()) * 100
```

```python
fig, ax1 = plt.subplots(figsize=(12, 6))

ax1.bar(pareto_data["PickupLocation"], pareto_data["BookingCount"], color="yellow", edgecolor="black", width=0.6)
ax1.set_xlabel("PickupLocation", fontsize=12, labelpad=10)
ax1.set_ylabel("Count of BookingID", color="black", fontsize=12)
ax1.tick_params(axis='y', labelcolor="black")
ax1.set_xticklabels(pareto_data["PickupLocation"], rotation=45, ha="right")

ax2 = ax1.twinx()
ax2.plot(pareto_data["PickupLocation"], pareto_data["CumulativePercentage"], color="darkorange", marker="o", ms=6, linewidth=2)
ax2.set_ylabel("Cumulative Percentage", color="darkorange", fontsize=12)
ax2.tick_params(axis='y', labelcolor="darkorange")

ax2.yaxis.set_major_formatter(PercentFormatter())

ax1.set_ylim(0, pareto_data["BookingCount"].max() * 1.1)
ax2.set_ylim(0, 105)

ax2.axhline(80, color="red", linestyle="--", alpha=0.7, label="80% Cut-off")

plt.title("Top Pickup Locations vs Ride Demand", fontsize=14, pad=15)

plt.tight_layout()
```
